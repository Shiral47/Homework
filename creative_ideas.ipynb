# Creative Ideas for Everyday Objects

This beginner-friendly notebook asks OpenAI for one unusual, practical way to use an everyday object. Run the cells from top to bottom in Google Colab.

## 1. Install the OpenAI SDK if needed

The next cell checks whether the SDK is already available and installs it only when necessary.

```python
# Check for the OpenAI package. Install it only if it is missing.
try:
    import openai  # noqa: F401
    print('OpenAI SDK is already installed.')
except ImportError:
    !pip -q install openai

from openai import OpenAI
from getpass import getpass
```

## 2. Enter your API key securely

Get an API key from your OpenAI account, then enter it when prompted. `getpass` hides the key as you type. Do not paste your key into a notebook cell or save it in the notebook.

```python
# The key is requested at runtime and is not printed or saved in this notebook.
api_key = getpass('Enter your OpenAI API key: ')
client = OpenAI(api_key=api_key)
```

## 3. Choose an everyday object

Enter an object you have around you, such as a paperclip, mug, or shoelace.

```python
everyday_object = input('What everyday object should we rethink? ').strip()

if not everyday_object:
    raise ValueError('Please enter an everyday object before continuing.')
```

## 4. Generate one creative idea

The model will suggest exactly one unusual, creative, and practical use for your object.

```python
response = client.responses.create(
    model='gpt-6-luna',
    input=(
        f'Suggest ONE unusual, creative, and practical use for a {everyday_object}. '
        'Explain briefly how to try it.'
    ),
)

print(response.output_text)
```
